# SymboLM — 03: Stage 1 Supervised Fine-Tuning (SFT)

Cold-start fine-tuning to teach the base model to think in the SymboLM symbolic DSL.
- 4-bit NF4 Quantization (QLoRA) fits comfortably inside 15GB VRAM on free Colab T4.
- Automatic checkpointing every 50 steps directly to `/content/drive/MyDrive/symboLM_checkpoints` so you never lose progress if Colab disconnects.

```python
# Run SFT training with automatic checkpointing and Google Drive sync
!python -m training.sft_train \
    --base_model Qwen/Qwen2.5-1.5B \
    --data_dir ./data/symbolic_dataset \
    --output_dir ./checkpoints/sft_adapter \
    --epochs 3 \
    --batch_size 2 \
    --grad_accum 8 \
    --lr 0.0002 \
    --drive_backup /content/drive/MyDrive/symboLM_checkpoints
```
