### 3. Interactive Playground & English Decompression

```python
from inference.generate import load_symbo_model, generate_solution
from inference.symbolic_to_text import decompress_trace

model, tokenizer = load_symbo_model("Qwen/Qwen2.5-1.5B", "./checkpoints/grpo_adapter")

question = "Lisa has 12 markers. She gives half to her brother, then buys a 6-pack. How many markers does she have now?"
print("Question:", question)

gen_output, ans = generate_solution(model, tokenizer, question, temperature=0.1, stream=False)
print("\n--- Model Output (Symbolic) ---")
print(gen_output)
print("\nFinal Answer:", ans)

print("\n--- Human-Readable Translation ---")
print(decompress_trace(gen_output))
```
