# SymboLM — 04: Stage 2 DeepSeek R1-Style GRPO Reinforcement Learning

Here we run **Group Relative Policy Optimization (GRPO)** on top of the SFT model:
- Generates $G=4$ completions per question
- Evaluates rewards: **Correctness** (+1.0/-0.5) + **Efficiency Token Compression** (+0 to +0.3) + **Format** (+0.1)
- Normalizes advantages within each group without requiring a separate critic network (saving 50% VRAM on T4!)
- Checkpoints automatically backed up to Google Drive.
